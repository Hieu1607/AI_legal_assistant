```python
# Use gemini-embedding-exp-03-07
import google.generativeai as genai
from dotenv import load_dotenv
load_dotenv()

genai.configure(api_key=os.getenv('Gemini_API_KEY')) # type: ignore

start_time = time.time()
result1 = genai.embed_content( # type: ignore
    model="models/gemini-embedding-exp-03-07",
    content=sentence1,
    task_type="RETRIEVAL_QUERY" # optimized for encoding short queries
)

result2 = genai.embed_content( # type: ignore
    model="models/gemini-embedding-exp-03-07",
    content=sentence2,
    task_type="RETRIEVAL_QUERY"
)
end_time = time.time()
run_time = end_time-start_time

# Extract embeddings from the result dictionaries
embedding1 = result1['embedding']
embedding2 = result2['embedding']

similarity = util.cos_sim(embedding1, embedding2)

print('Runtime: ',run_time)
print('Embedding of sentence 1: ', embedding1)
print('Embedding of sentence 2: ', embedding2)
print('Number of dimensions: ', len(embedding1))
print('Cosine similarity: ', similarity.item())
```

```python
# Use gemini-embedding-001
import google.generativeai as genai
from dotenv import load_dotenv
load_dotenv()

GOOGLE_API_KEY = os.getenv('Gemini_API_KEY')
genai.configure(api_key=GOOGLE_API_KEY) # type: ignore
start_time = time.time()

result1 = genai.embed_content( # type: ignore
    model="models/embedding-001",
    content=sentence1,
    task_type="RETRIEVAL_QUERY"
)

result2 = genai.embed_content( # type: ignore
    model="models/embedding-001",
    content=sentence2,
    task_type="RETRIEVAL_QUERY"
)
end_time = time.time()

# Extract embeddings from the result dictionaries
embedding1 = result1['embedding']
embedding2 = result2['embedding']

similarity = util.cos_sim(embedding1, embedding2)

print('Runtime: ',end_time-start_time)
print('Embedding of sentence 1: ', embedding1)
print('Embedding of sentence 2: ', embedding2)
print('Number of dimensions: ', len(embedding1))
print('Cosine similarity: ', similarity.item())
```

```python
# Use all-MiniLM-L6-v2
from sentence_transformers import SentenceTransformer, util


sentence1 = "Tôi muốn biết về hợp đồng lao động."
sentence2 = "Hợp đồng lao động là gì?"

start_time = time.time()
# Load model embedding
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')


embedding1 = model.encode(sentence1, normalize_embeddings=True)
embedding2 = model.encode(sentence2, normalize_embeddings=True)
end_time = time.time()

similarity = util.cos_sim(embedding1, embedding2)

print('Runtime: ',end_time-start_time)
print('Embedding of sentence 1: ', embedding1)
print('Embedding of sentence 2: ', embedding2)
print('Number of dimensions: ', len(embedding1))
print("Cosine Similarity:", similarity.item())
```
